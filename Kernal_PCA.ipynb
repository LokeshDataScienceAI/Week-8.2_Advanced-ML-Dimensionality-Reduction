**IMPORTING LIBRARIES**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import KernelPCA
from sklearn.metrics import r2_score
```

**FEATURE EXTRACTION - KERNEL PCA**

```python
def kernel_pca(indep_X, n):
    kpca = KernelPCA(
        n_components=n,
        kernel='rbf',
        random_state=0
    )

    X_kpca = kpca.fit_transform(indep_X)

    return X_kpca
```

**TRAIN TEST SPLIT & SCALING**

```python
def split_scalar(indep_X, dep_Y):

    X_train, X_test, y_train, y_test = train_test_split(
        indep_X,
        dep_Y,
        test_size=0.25,
        random_state=0
    )

    sc = StandardScaler()

    X_train = sc.fit_transform(X_train)
    X_test = sc.transform(X_test)

    return X_train, X_test, y_train, y_test
```

**R2 SCORE FUNCTION**

```python
def r2_prediction(regressor, X_test, y_test):

    y_pred = regressor.predict(X_test)

    r2 = r2_score(y_test, y_pred)

    return r2
```

**LINEAR REGRESSION**

```python
def Linear(X_train, y_train, X_test, y_test):

    from sklearn.linear_model import LinearRegression

    regressor = LinearRegression()

    regressor.fit(X_train, y_train)

    r2 = r2_prediction(regressor, X_test, y_test)

    return r2
```

**SVM LINEAR**

```python
def svm_linear(X_train, y_train, X_test, y_test):

    from sklearn.svm import SVR

    regressor = SVR(kernel='linear')

    regressor.fit(X_train, y_train)

    r2 = r2_prediction(regressor, X_test, y_test)

    return r2
```

**SVM RBF**

```python
def svm_NL(X_train, y_train, X_test, y_test):

    from sklearn.svm import SVR

    regressor = SVR(kernel='rbf')

    regressor.fit(X_train, y_train)

    r2 = r2_prediction(regressor, X_test, y_test)

    return r2
```

**DECISION TREE**

```python
def Decision(X_train, y_train, X_test, y_test):

    from sklearn.tree import DecisionTreeRegressor

    regressor = DecisionTreeRegressor(random_state=0)

    regressor.fit(X_train, y_train)

    r2 = r2_prediction(regressor, X_test, y_test)

    return r2
```

**RANDOM FOREST**

```python
def Random(X_train, y_train, X_test, y_test):

    from sklearn.ensemble import RandomForestRegressor

    regressor = RandomForestRegressor(
        n_estimators=100,
        random_state=0
    )

    regressor.fit(X_train, y_train)

    r2 = r2_prediction(regressor, X_test, y_test)

    return r2
```

```python
def create_result_table(acclin,
                        accsvml,
                        accsvmnl,
                        accdes,
                        accrf):

    dataframe = pd.DataFrame(
        index=['Kernel PCA'],
        columns=[
            'Linear',
            'SVM Linear',
            'SVM RBF',
            'Decision Tree',
            'Random Forest'
        ]
    )

    dataframe.loc['Kernel PCA', 'Linear'] = acclin
    dataframe.loc['Kernel PCA', 'SVM Linear'] = accsvml
    dataframe.loc['Kernel PCA', 'SVM RBF'] = accsvmnl
    dataframe.loc['Kernel PCA', 'Decision Tree'] = accdes
    dataframe.loc['Kernel PCA', 'Random Forest'] = accrf

    return dataframe
```

**DATASET LOADING**

```python
dataset = pd.read_csv("prep.csv")
```

```python
dataset
```

**DATA PREPROCESSING**

```python
dataset = pd.get_dummies(
    dataset,
    drop_first=True
)
```

```python
dataset
```

**INDEPENDENT & DEPENDENT**

```python
indep_X = dataset.drop(
    'classification_yes',
    axis=1
)

dep_Y = dataset['classification_yes']
```

**APPLY KERNEL PCA**

```python
X_kpca = kernel_pca(
    indep_X,
    n=5
)
```

**SPLIT DATA**

```python
X_train, X_test, y_train, y_test = split_scalar(
    X_kpca,
    dep_Y
)
```

**MODEL TRAINING**

```python
acclin = Linear(
    X_train,
    y_train,
    X_test,
    y_test
)

accsvml = svm_linear(
    X_train,
    y_train,
    X_test,
    y_test
)

accsvmnl = svm_NL(
    X_train,
    y_train,
    X_test,
    y_test
)

accdes = Decision(
    X_train,
    y_train,
    X_test,
    y_test
)

accrf = Random(
    X_train,
    y_train,
    X_test,
    y_test
)
```

**RESULT TABLE**

```python
result = create_result_table(
    acclin,
    accsvml,
    accsvmnl,
    accdes,
    accrf
)
```

```python
print("\nKernel PCA Model Comparison\n")

print(result)

# Optional
result.to_csv("KernelPCA_Result.csv")
```
